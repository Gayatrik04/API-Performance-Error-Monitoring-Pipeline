```python
import numpy as np
import pandas as pd
```

```python
api_usage=pd.read_excel("api_usage_5k_dirty_dataset.xlsx")
```

```python
api_usage.info()
```

```python
api_usage['user_id']=api_usage['user_id'].fillna(0)
api_usage['api_id']=api_usage['api_id'].fillna(0)

api_usage['user_id'].isnull().sum()
api_usage['api_id'].isnull().sum()

api_usage['user_id']=api_usage['user_id'].astype(int)
api_usage['api_id']=api_usage['api_id'].astype(int)
```

```python
api_usage['usage_date']=pd.to_datetime(api_usage['usage_date'],format="mixed",errors='coerce')

api_usage['usage_date']=api_usage['usage_date'].fillna(api_usage['usage_date'].mode()[0])
api_usage['usage_date'].isnull().sum()
```

```python
api_usage['total_requests']=api_usage['total_requests'].fillna(0)
api_usage['successful_requests']=api_usage['successful_requests'].fillna(0)
api_usage['failed_requests']=api_usage['failed_requests'].fillna(0)

api_usage['total_requests']=api_usage['total_requests'].astype(int)
api_usage['successful_requests']=api_usage['successful_requests'].astype(int)
api_usage['failed_requests']=api_usage['failed_requests'].astype(int)
```

```python
api_usage.info()
```

```python
api_usage.describe()
```

```python
api_usage['successful_requests'] = api_usage['successful_requests'].clip(lower=0)

api_usage['failed_requests'] = api_usage['failed_requests'].clip(lower=0)
```

```python
api_usage['total_requests'] = (
    api_usage['successful_requests'] +
    api_usage['failed_requests']
)
```

```python
api_usage.info()
```

```python
api_usage.describe()
```

```python
api_usage[(api_usage['user_id'] == 0) | (api_usage['api_id'] == 0)]
api_usage = api_usage[
    (api_usage['user_id'] != 0) & (api_usage['api_id'] != 0)
]
```

```python
api_usage.describe()
```

```python
api_usage[(api_usage['user_id'] == 0) | (api_usage['api_id'] == 0)]
api_usage.isnull().sum()
api_usage.duplicated().sum()


(api_usage['total_requests']==
    api_usage['successful_requests']+ api_usage['failed_requests']
).all()

api_usage['total_requests']=api_usage['successful_requests']+api_usage['failed_requests']
```

```python
api_usage.describe()
```

```python
api_usage['success_rate'] = (
  api_usage['successful_requests']/ api_usage['total_requests']) * 100

api_usage['success_rate'] = api_usage['success_rate'].fillna(0)
```

```python
api_usage['error_rate'] = (
  api_usage['failed_requests']/ api_usage['total_requests']) * 100

api_usage['error_rate'] = api_usage['error_rate'].fillna(0)
```

```python
api_usage.to_csv("cleaned_api_usage.csv",index=False)
```
