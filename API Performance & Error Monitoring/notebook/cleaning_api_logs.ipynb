```python
import numpy as np
import pandas as pd
```

```python
api_logs=pd.read_excel("api_logs_5k_dirty_dataset.xlsx")
```

```python
api_logs.info()
```

```python
api_logs['request_time'] = pd.to_datetime(
  api_logs['request_time'],format='mixed',errors='coerce')
```

```python
api_logs['request_time'] = api_logs['request_time'].fillna(
  api_logs['request_time'].mode()[0])
```

```python
api_logs['response_time_ms'] = api_logs['response_time_ms'].fillna(
  api_logs['response_time_ms'].median())
```

```python
api_logs['cpu_usage'] = api_logs['cpu_usage'].fillna(
  api_logs['cpu_usage'].median()
)

api_logs['memory_usage'] = api_logs['memory_usage'].fillna(
  api_logs['memory_usage'].median()
)
```

```python
api_logs.info()
```

```python
api_logs['error_message'] = api_logs['error_message'].fillna('No Error')
```

```python
api_logs['request_method'].unique()
api_logs['request_method'] = (
  api_logs['request_method'].str.strip().str.upper())
api_logs['request_method'].unique()
```

```python
api_logs['server_region'].unique()
api_logs['server_region'] = (api_logs['server_region'].str.strip().str.title())
api_logs['server_region'].unique()
```

```python
api_logs['is_error'] = api_logs['status_code'] >= 400
```

```python
def performance_category(x):
    if x < 200:
        return 'Fast'
    elif x < 500:
        return 'Moderate'
    else:
        return 'Slow'

api_logs['performance_category'] = api_logs['response_time_ms'].apply(performance_category)
```

```python
api_logs.isnull().sum()
api_logs.info()
```

```python
api_logs.to_csv('cleaned_api_logs.csv',index=False)
```
