```python
import pandas as pd
import numpy as np
```

```python
users=pd.read_excel("users_5k_dirty_dataset.xlsx")
```

```python
users.info()
```

```python
users['user_id'].isnull().sum()
users['user_id'].duplicated().sum()
```

```python
users['username'].isnull().sum()
users['username'].duplicated().sum()
users['username']=(users['username'].str.strip().str.lower())
users[users.duplicated(subset=['username'],keep=False)]
```

```python
users['company_name'].isnull().sum()
users['company_name']=(users['company_name'].fillna('Unknowm').str.strip())
users['company_name'].isnull().sum()
users['company_name'].unique()
```

```python
users['subscription_type'].isnull().sum()
users['subscription_type'].unique()
users['subscription_type']=(users['subscription_type'].str.strip().str.lower().fillna("Unknown"))
users['subscription_type'].unique()
users['subscription_type']=(users['subscription_type'].replace({
    'free': 'Free',
    'basic': 'Basic',
    'premium': 'Premium',
    'enterprise': 'Enterprise',
    'corp': 'Corporate',
    'corporate': 'Corporate',
    'gold': 'Gold',
    'standard': 'Standard',
    'unknown': 'Unknown'
}))

users['subscription_type'].unique()
```

```python
users['country'].isnull().sum()
users['country']=(users['country'].fillna('Unknown').str.strip().str.title())
users['country'].unique()
users['country'] = users['country'].replace({
    'Us': 'United States',
    'U.S.A.': 'United States',

    'Uk': 'United Kingdom',
    'U.K.': 'United Kingdom',

    'De': 'Germany',
    'Deutschland': 'Germany'
})
users['country'].unique()
```

```python
users['signup_date']=pd.to_datetime(users['signup_date'],format='mixed',errors='coerce')
users['signup_date'].isnull().sum()

users['signup_date'] = users['signup_date'].fillna(
    users['signup_date'].mean())
users['signup_date'].isnull().sum()
```

```python
users.info()
```

```python
users.to_csv('cleaned_users_dataset.csv',index=False)
```
